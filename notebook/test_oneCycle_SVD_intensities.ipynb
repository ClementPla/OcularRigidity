```python
# load videos
def load_time_vector_from_xml(xml_path):
    """
    Parameters
    ----------
    xml_path : str
        Path to the Spectralis (HEYEX) XML export.

    Returns
    -------
    time : (N,) float64 ndarray
        Elapsed time in seconds, with the first sample at t = 0.
    """
    study = SpectralisStudy.from_file(xml_path)
    time = np.array([t.seconds_of_day for t in study.timestamps], dtype=np.float64)
    time = time - time.min()
    return time


path_condi = Path("E:/SANSORI/01_210713001/210713001before_rigidity/210713001before_rigidity_OD")
path_video = Path(path_condi / "RawImages" / "registered" / "registered_video_ascan.mp4")
path_mask = Path(path_condi / "RawImages" / "registered" / "mask_ascan.npz")
path_idx2keep = Path(path_condi / "RawImages" / "registeredBscans" / "detection_outliers_median_threshold.mat")
path_heartbeat = Path(path_condi / "Data Files" / "visit_data.csv")
path_xml = sorted((path_condi / "RawImages").glob("*.xml"))[0]

video = read_gray(path_video)
mask = load_mask(path_mask)
mask_intersection = mask.all(axis=0)  # intersection of all masks
t = load_time_vector_from_xml(path_xml)
ids = loadmat(path_idx2keep)["idsFilteredMmi"].ravel()
t_filt = t[ids.astype(int) - 1]

print(f"Video shape: {video.shape}")
print(f"Mask shape: {mask.shape}")
print(f"Mask shape: {mask_intersection.shape}")
print(f"Time vector shape: {t.shape}")
print(f"Filtered time vector shape: {t_filt.shape}")

plt.imshow(mask_intersection, cmap="gray")
```
